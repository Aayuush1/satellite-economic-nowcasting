```python
import ee
import pandas as pd

# Initialize Earth Engine
ee.Initialize(project='satellite-economic-nowcasting')

# Use FAO GAUL 2015 with standard property matching
# (Deprecation warning is purely informational; 2015 remains fully functional)
try:
    nigeria_states = (
        ee.FeatureCollection("FAO/GAUL/2015/level1")
        .filter(ee.Filter.eq('ADM0_NAME', 'Nigeria'))
    )
    nigeria_lgas = (
        ee.FeatureCollection("FAO/GAUL/2015/level2")
        .filter(ee.Filter.eq('ADM0_NAME', 'Nigeria'))
    )
    
    num_states = nigeria_states.size().getInfo()
    num_lgas = nigeria_lgas.size().getInfo()
except Exception as e:
    print(f"Error loading boundaries: {e}")

print(f"Loaded Nigeria Admin-1 (States): {num_states} units")
print(f"Loaded Nigeria Admin-2 (LGAs):   {num_lgas} units")
```

```python
# Convert state attributes to a Pandas DataFrame
state_features = nigeria_states.reduceColumns(
    reducer=ee.Reducer.toList(2),
    selectors=['ADM1_CODE', 'ADM1_NAME']
).getInfo()['list']

df_states = pd.DataFrame(state_features, columns=['state_code', 'state_name']).sort_values('state_name')
print("Sample States:")
print(df_states.head(10))

# Save local copy of the administrative index
df_states.to_csv('../data/processed/nigeria_states_index.csv', index=False)
print("\nSaved state index to data/processed/nigeria_states_index.csv")
```
